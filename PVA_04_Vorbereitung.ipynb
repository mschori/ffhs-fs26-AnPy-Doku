# Integralrechnung

Die Integralrechnung ist die Umkehroperation der Differentialrechnung.

Während beim Ableiten gefragt wird:

> „Wie verändert sich eine Funktion?“

fragt die Integralrechnung:

> „Welche Funktion wurde ursprünglich abgeleitet?“

Zusätzlich kann mit Integralen ein Flächeninhalt berechnet werden.

---

# Erläuterung von Begriffen

- das $d$ in $dx$ steht für „Differential von x“ und zeigt die Variable an, über die integriert wird
- --> also ist $d$ ein wizig kleines Stück der Variable --> Beispiel: $\int f8x) dx$ bedeutet:
- - $f(x)$ = die Funktion
- - $dx$ = Integration bezüglich $x$
- - man liest also: Integriere $f(x)$ nach $x$
- die Konstante $C$ in der unbestimmten Integration steht für eine beliebige Konstante, da die Ableitung einer Konstanten Null ist
- die Grenzen $a$ und $b$ in der bestimmten Integration geben den Bereich an, über den die Fläche berechnet wird
- die Funktion $f(x)$ wird als Integrand bezeichnet, da sie die Funktion ist, die integriert wird

# 1. Stammfunktion und unbestimmtes Integral

## Definition Stammfunktion

Eine Funktion $F(x)$ heißt Stammfunktion von $f(x)$, wenn gilt:

$$
F'(x) = f(x)
$$

Das bedeutet:

- $F(x)$ ist die ursprüngliche Funktion
- $f(x)$ ist ihre Ableitung

---

## Beispiel

Gegeben:

$$
f(x) = 3x^2
$$

Gesucht ist eine Stammfunktion.

Da gilt:

$$
(x^3)' = 3x^2
$$

ist eine Stammfunktion:

$$
F(x) = x^3
$$

Aber auch:

$$
F(x) = x^3 + 5
$$

denn:

$$
(x^3 + 5)' = 3x^2
$$

Die Konstante verschwindet beim Ableiten.

---

## Unbestimmtes Integral

Die Menge aller Stammfunktionen schreibt man als:

$$
\int f(x)\,dx = F(x) + C
$$

Dabei bedeutet:

| Begriff | Bedeutung |
|---|---|
| $f(x)$ | Integrand |
| $dx$ | Integrationsvariable |
| $C$ | Integrationskonstante |

---



# 2. Wichtige Stammfunktionen

## Konstanten

$$
\int c\,dx = cx + C
$$

---

## Potenzregel

Für $a \neq -1$:

$$
\int x^a\,dx = \frac{x^{a+1}}{a+1} + C
$$

---

## Spezialfall $1/x$

$$
\int \frac{1}{x}\,dx = \ln|x| + C
$$

---

## Exponentialfunktionen

$$
\int e^x\,dx = e^x + C
$$

$$
\int a^x\,dx = \frac{a^x}{\ln(a)} + C
$$

---

## Trigonometrische Funktionen

$$
\int \sin(x)\,dx = -\cos(x) + C
$$

$$
\int \cos(x)\,dx = \sin(x) + C
$$

---



# 3. Linearität der Integration

Integrale dürfen gliedweise berechnet werden.

## Summe

$$
\int (f(x)+g(x))\,dx
=
\int f(x)\,dx + \int g(x)\,dx
$$

---

## Konstanter Faktor

$$
\int k \cdot f(x)\,dx
=
k \int f(x)\,dx
$$

---

## Beispiel

Berechne:

$$
\int (2x^3 + 4x)\,dx
$$

Das besteht aus zwei Teilen: $2x^3$ und $4x$. Das darf aufgeteilt werden.

Aufteilen:

$$
\int 2x^3\,dx + \int 4x\,dx
$$

Regel für konstante Faktoren ($\int k * f(x) dx = k \int f(x) dx$) anwenden:

$$
=
2\int x^3\,dx + 4\int x\,dx
$$

Einsetzen der einzelnen Integrale (Potenzregel: $\int x^a dx = \frac{x^{a+1}}{a+1} + C$):

$$
=
2 \cdot \frac{x^4}{4} + 4 \cdot \frac{x^2}{2} + C
$$

Vereinfachen:

$$
=
\frac{x^4}{2} + 2x^2 + C
$$

---



# 4. Integration durch Substitution

Die Substitution basiert auf der Kettenregel.

Sie wird verwendet, wenn eine Funktion „innen“ und deren Ableitung ebenfalls vorkommen.

---

## Grundidee

Wenn der Integrand die Form hat:

$$
f(g(x)) \cdot g'(x)
$$

setzt man:

$$
u = g(x)
$$

Dann wird:

$$
du = g'(x)\,dx
$$

---

## Beispiel

Berechne:

$$
\int \sin(x^2)\cdot 2x\,dx
$$

Substitution:

$$
u = x^2
$$

Dann:

$$
du = 2x\,dx
$$

Integral wird zu:

$$
\int \sin(u)\,du
$$

Integrieren:

$$
=
-\cos(u) + C
$$

Rücksubstitution:

$$
=
-\cos(x^2) + C
$$

---



# 5. Partielle Integration

Die partielle Integration basiert auf der Produktregel.

---

## Formel

$$
\int f(x)g'(x)\,dx
=
f(x)g(x) - \int f'(x)g(x)\,dx
$$

---

## Typischer Einsatz

Bei Produkten wie:

- $x e^x$
- $x \sin(x)$
- $x \ln(x)$

---

## Beispiel

Berechne:

$$
\int x e^x\,dx
$$

Setze:

$$
f(x)=x
$$

$$
g'(x)=e^x
$$

Dann:

$$
f'(x)=1
$$

$$
g(x)=e^x
$$

Einsetzen:

$$
=
xe^x - \int e^x\,dx
$$

$$
=
xe^x - e^x + C
$$

---



# 6. Bestimmtes Integral

Das bestimmte Integral berechnet den orientierten Flächeninhalt zwischen Funktion und x-Achse.

---

## Schreibweise

$$
\int_a^b f(x)\,dx
$$

| Symbol | Bedeutung |
|---|---|
| $a$ | untere Grenze |
| $b$ | obere Grenze |
| $f(x)$ | Integrand |

---

## Geometrische Bedeutung

- Fläche oberhalb der x-Achse → positiv
- Fläche unterhalb der x-Achse → negativ

---



# 7. Hauptsatz der Differential- und Integralrechnung

Der wichtigste Satz der Integralrechnung:

Wenn $F(x)$ eine Stammfunktion von $f(x)$ ist:

$$
\int_a^b f(x)\,dx
=
F(b)-F(a)
$$

---

## Vorgehen

1. Stammfunktion bilden
2. Obere Grenze einsetzen
3. Untere Grenze einsetzen
4. Subtrahieren

---

## Beispiel

Berechne:

$$
\int_0^1 x^2\,dx
$$

Stammfunktion:

$$
F(x)=\frac{x^3}{3}
$$

Einsetzen:

$$
=
\frac{1^3}{3} - \frac{0^3}{3}
$$

$$
=
\frac13
$$

---



# 8. Bestimmte Integration mit Substitution

## Beispiel

Berechne:

$$
\int_0^\pi \sin(x^2)x\,dx
$$

Substitution:

$$
u=x^2
$$

Dann:

$$
du=2x\,dx
$$

Also:

$$
x\,dx = \frac12 du
$$

Grenzen umrechnen:

- $x=0 \Rightarrow u=0$
- $x=\pi \Rightarrow u=\pi^2$

Neues Integral:

$$
\frac12 \int_0^{\pi^2}\sin(u)\,du
$$

---



# 9. Uneigentliche Integrale

Uneigentliche Integrale entstehen:

## Fall 1 – Unendliche Grenze

$$
\int_1^\infty \frac1{x^2}\,dx
$$

---

## Fall 2 – Unbeschränkter Integrand

$$
\int_0^1 \frac1{\sqrt{x}}\,dx
$$

---

## Definition über Grenzwerte

Beispiel:

$$
\int_1^\infty \frac1{x^2}\,dx
=
\lim_{c\to\infty}
\int_1^c \frac1{x^2}\,dx
$$

---



# 10. Konvergenz und Divergenz

## Konvergent

Das Integral besitzt einen endlichen Wert.

Beispiel:

$$
\int_1^\infty \frac1{x^2}\,dx = 1
$$

---

## Divergent

Das Integral wächst unbegrenzt.

Beispiel:

$$
\int_1^\infty \frac1x\,dx
=
\infty
$$

---



# 11. Numerische Integration

Nicht jedes Integral kann analytisch gelöst werden.

Dann verwendet man numerische Verfahren.

---

## Trapezregel

Approximation mit Trapezen.

Formel:

$$
\int_a^b f(x)\,dx
\approx
\frac{b-a}{2}(f(a)+f(b))
$$

---

## Simpsonregel

Approximation mit Parabeln.

Genauer als die Trapezregel.

---



# 12. Python mit SymPy

## Vorbereitung

```python
import sympy as sp

````python
# Vorbereitung
import sympy as sp

# Variablen definieren
x = sp.symbols('x')

# Unbestimmtes Integral
sp.integrate(x**2, x)

# Ergebnis
x**3 / 3

# Bestimmtes Integral
sp.integrate(x**2, (x, 0, 1))

# Ergebnis
1/3

# Beispiel mit Sinus
sp.integrate(sp.sin(x), x)

# Numerische Auswertung
sp.N(sp.integrate(sp.sin(x), (x, 0, sp.pi)))

````

# 13. Numerische Integration it SciPY


````python
# Vorbereitung
import numpy as np
from scipy import integrate

# Funktion definieren
def f(x):
    return x**2

# Integral berechnen
result, error = integrate.quad(f, 0, 1)

print(result)

# Ergebnis
0.3333333333333333

# Beispiel mit Sinus
def f(x):
    return np.sin(x)

result, error = integrate.quad(f, 0, np.pi)

print(result)

# Ergebnis
2.0
````
