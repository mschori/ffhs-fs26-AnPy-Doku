## Einleitung

Die Approximation von Funktionen ist ein zentraler Bestandteil der numerischen Mathematik und spielt in vielen technischen und wissenschaftlichen Anwendungen eine wichtige Rolle. In der Praxis sind exakte Funktionsverläufe häufig nicht direkt bekannt, sondern nur durch eine endliche Anzahl von Mess- oder Stützpunkten gegeben. Um dennoch Aussagen über den Verlauf zwischen diesen Punkten machen zu können, werden Approximationsverfahren eingesetzt.

Eine besonders wichtige Methode zur Beschreibung glatter Kurven zwischen gegebenen Punkten sind Splines. Dabei handelt es sich um stückweise definierte Funktionen, die so miteinander verbunden werden, dass ein gleichmässiger und glatter Kurvenverlauf entsteht [1]. Kubische Splines sind in diesem Zusammenhang besonders relevant, da sie eine gute Balance zwischen mathematischer Einfachheit, Glattheit und Approximationsqualität bieten [2].

Ziel dieser Arbeit ist es, eine nichtlineare Kurve anhand ausgewählter Stützstellen mit kubischen Splines zu approximieren. Als Beispiel wird die Funktion

$$f(x)=sin(x)+0.3x$$

verwendet. Diese Funktion eignet sich gut für die Untersuchung, da sie einerseits einen trigonometrischen Anteil besitzt und damit die Vorgaben der Aufgabenstellung erfüllt, andererseits durch den linearen Zusatz einen ansteigenden Trend aufweist. Dadurch lassen sich Unterschiede zwischen exakter Funktion und Approximation besser erkennen als bei einer reinen Sinusfunktion.

Im Zentrum der Arbeit stehen die grafische Darstellung der exakten Funktion, der Spline-Approximation und der verwendeten Stützpunkte sowie die Diskussion der Approximationsqualität. Insbesondere wird untersucht, wie sich die Anzahl der Stützstellen auf die Genauigkeit der Approximation auswirkt und welche Möglichkeiten zur Verbesserung der Approximationsgüte bestehen.

## Theoretische Grundlagen
Splines sind stückweise definierte Funktionen, die zur Interpolation oder Approximation gegebener Datenpunkte verwendet werden. Anstatt eine einzige Funktion über das gesamte Intervall zu bestimmen, wird das Intervall in mehrere Teilintervalle zerlegt. Für jedes dieser Teilintervalle wird ein eigenes Polynom definiert. Diese Teilfunktionen werden anschliessend so miteinander verbunden, dass ein zusammenhängender und möglichst glatter Gesamtverlauf entsteht.

Grundsätzlich kann zwischen Interpolation und Approximation unterschieden werden [2]. Bei der Interpolation verläuft die berechnete Funktion exakt durch alle gegebenen Stützpunkte. Bei der Approximation hingegen wird lediglich versucht, den Funktionsverlauf möglichst gut nachzubilden, ohne zwingend jeden Punkt exakt zu treffen. In dieser Arbeit wird eine interpolierende Methode verwendet, da die Splines die gegebenen Stützstellen exakt durchlaufen [1], [4].

In dieser Arbeit werden kubische Splines eingesetzt. Dabei handelt es sich um Polynome dritten Grades, die zwischen jeweils zwei benachbarten Stützstellen definiert sind [1], [5]. Kubische Splines werden so konstruiert, dass sie an den Verbindungsstellen nicht nur denselben Funktionswert besitzen, sondern auch in der ersten und zweiten Ableitung übereinstimmen [1], [5]. Dadurch entstehen keine Sprünge oder Knicke, sondern ein glatter Kurvenverlauf.

Die Stetigkeit der ersten Ableitung sorgt dafür, dass die Tangentenrichtung an den Übergangsstellen übereinstimmt. Die zusätzliche Stetigkeit der zweiten Ableitung stellt sicher, dass auch die Krümmung der Kurve keine abrupten Änderungen aufweist. Diese Eigenschaften sind insbesondere für Anwendungen relevant, in denen ein natürlicher und physikalisch plausibler Verlauf gefordert ist.

Kubische Splines stellen einen geeigneten Kompromiss dar [2]. Polynome niedrigeren Grades sind oft nicht flexibel genug, während Polynome höheren Grades zu instabilen Verläufen und unerwünschten Schwingungen neigen können (Runge-Phänomen). Durch die Verwendung von Polynomen dritten Grades wird eine ausreichende Flexibilität bei gleichzeitig hoher Stabilität erreicht.

Das sogenannte Runge-Phänomen beschreibt das Problem, dass globale Polynome höheren Grades zwischen den Stützstellen starke Schwingungen entwickeln können. Diese unerwünschten Oszillationen treten insbesondere bei einer grossen Anzahl von Stützstellen oder an den Randbereichen eines Intervalls auf. Splines vermeiden dieses Problem, da sie nicht als einzelnes globales Polynom definiert sind, sondern aus mehreren lokalen Teilfunktionen bestehen. Dadurch bleibt der Kurvenverlauf stabiler und besser kontrollierbar.

Ein wesentlicher Vorteil von Splines gegenüber globalen Polynomen besteht darin, dass sie lokal definiert sind. Änderungen an einzelnen Stützstellen wirken sich nur auf benachbarte Teilintervalle aus und nicht auf die gesamte Kurve. Dadurch wird die Methode robuster gegenüber lokalen Veränderungen.

Neben der Wahl der Stützstellen sind auch die Randbedingungen von Bedeutung. Da zusätzliche Bedingungen erforderlich sind, um den Spline eindeutig zu bestimmen, werden in dieser Arbeit sogenannte natürliche Randbedingungen verwendet [1], [4]. Dabei wird angenommen, dass die zweite Ableitung an den Randpunkten gleich null ist. Dies führt zu einem glatten und nicht erzwungenen Randverhalten der Kurve.

Die Qualität einer Spline-Approximation hängt wesentlich von der Anzahl und Verteilung der Stützstellen ab. Mit wenigen Stützstellen kann nur ein grober Verlauf erfasst werden, während eine höhere Anzahl eine genauere Approximation ermöglicht. Besonders in Bereichen mit starker Krümmung ist eine ausreichende Punktdichte erforderlich.

Zur Beurteilung der Qualität einer Approximation können sowohl visuelle als auch numerische Methoden verwendet werden [2]. Die grafische Darstellung ermöglicht eine intuitive Einschätzung, während die Berechnung von Fehlergrössen, wie dem maximalen oder mittleren Fehler, eine quantitative Bewertung liefert.

Durch die Wahl der Funktion $f(x)=sin(x)+0.3x$ ergibt sich ein anschauliches Beispiel. Die Funktion kombiniert periodisches Verhalten mit einem linearen Trend und weist unterschiedliche Krümmungsbereiche auf. Dadurch eignet sie sich besonders gut, um den Einfluss der Stützstellen auf die Approximation sichtbar zu machen.

## Implementierungsidee

Die Umsetzung basiert auf folgenden Komponenten:

- Definition der exakten Funktion
- Erzeugung von Stützstellen mit unterschiedlicher Anzahl
- Berechnung der Spline-Approximation mit CubicSpline [4]
- Visualisierung der Ergebnisse mit Matplotlib
- Einfache Fehleranalyse durch Vergleich von exakter Funktion und Approximation

Zur Untersuchung des Einflusses der Stützstellen werden drei Varianten betrachtet:

- 4 Stützstellen (grobe Approximation)
- 8 Stützstellen (mittlere Approximation)
- 12 Stützstellen (feinere Approximation)

Für jede Variante wird eine separate Grafik erstellt, die folgende Elemente enthält:

- exakte Funktion
- Spline-Approximation
- Stützpunkte

## Programmcode
### Imports

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import CubicSpline
```

## Definition der exakten Funktion

```python
def exact_function(x: np.ndarray) -> np.ndarray:
    """
    Exact function used for comparison.
    """
    return np.sin(x) + 0.3 * x
```

### Erstellung der Stützstellen

```python
def create_support_points(n: int, start: float, end: float):
    """
    Creates n support points in a given interval.
    """
    x = np.linspace(start, end, n)
    y = exact_function(x)
    return x, y
```

### Erstellung der Splines

```python
def create_spline(x: np.ndarray, y: np.ndarray) -> CubicSpline:
    """
    Creates a cubic spline with natural boundary conditions.
    The implementation is based on SciPy's CubicSpline class [4].
    """
    return CubicSpline(x, y, bc_type='natural')
```

### Plot-Funktion

```python
def plot_spline_approximation(n_points: int, start: float, end: float) -> None:
    """
    Plots the exact function, the spline approximation and the support points.
    """
    x_support, y_support = create_support_points(n_points, start, end)
    spline = create_spline(x_support, y_support)

    x_fine = np.linspace(start, end, 500)
    y_exact = exact_function(x_fine)
    y_spline = spline(x_fine)

    error = np.abs(y_exact - y_spline)
    max_error = np.max(error)

    plt.figure()
    plt.style.use('ggplot')
    plt.plot(x_fine, y_exact, label="Exakte Funktion sin(x) + 0.3x")
    plt.plot(x_fine, y_spline, linestyle="--", label="Spline Approximation")
    plt.scatter(x_support, y_support, color="red", label="Stützstellen")

    plt.title(f"Spline Approximation mit {n_points} Stützstellen "
              f"(max. Fehler: {max_error:.4f})")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)
    plt.show()
```

## Darstellung der Ergebnisse
Zur Analyse der Approximation wird die Funktion $f(x)=sin(x)+0.3x$ im Intervall [−2,2π] betrachtet. Die grafische Darstellung ermöglicht eine direkte visuelle Beurteilung der Approximationsqualität und macht Unterschiede zwischen den Varianten gut erkennbar.

### Variante 1: 4 Stützstellen
Die Approximation mit 4 Stützstellen zeigt deutliche Abweichungen vom exakten Funktionsverlauf. Aufgrund der grossen Abstände zwischen den einzelnen Punkten kann der Spline die Krümmung der Funktion nur unzureichend erfassen. Insbesondere in Bereichen mit stärkerer Änderung des Funktionsverlaufs sind sichtbare Unterschiede erkennbar. Die Approximation wirkt teilweise zu flach und weicht zwischen den Stützstellen deutlich von der exakten Funktion ab. Auch die Fehleranalyse zeigt eine vergleichsweise grosse maximale Abweichung.

```python
plot_spline_approximation(4, -2, 2 * np.pi)
```

### Variante 2: 8 Stützstellen
Mit 8 Stützstellen verbessert sich die Approximation deutlich. Die Abstände zwischen den Punkten sind kleiner, wodurch der Spline den Funktionsverlauf genauer nachbilden kann. Die Abweichungen sind weiterhin vorhanden, jedoch deutlich geringer als bei der ersten Variante. Besonders in Bereichen mit moderater Krümmung stimmt der Verlauf bereits gut mit der exakten Funktion überein. Die maximale Abweichung reduziert sich entsprechend deutlich.

```python
plot_spline_approximation(8, -2, 2 * np.pi)
```

### Variante 3: 12 Stützstellen
Bei 12 Stützstellen ist die Approximation nahezu identisch mit der exakten Funktion. Die Unterschiede sind visuell kaum noch erkennbar. Der Spline kann sowohl den linearen Trend als auch die periodischen Anteile der Funktion sehr präzise abbilden. Auch die Fehleranalyse zeigt, dass die maximale Abweichung nur noch gering ist.

```python
plot_spline_approximation(12, -2, 2 * np.pi)
```

## Diskussion der Ergebnisse
Die Ergebnisse zeigen deutlich, dass die Anzahl und Verteilung der Stützstellen einen entscheidenden Einfluss auf die Qualität der Spline-Approximation haben [1], [2].

Bei der Verwendung von nur 4 Stützstellen ist die Approximation vergleichsweise ungenau. Die grossen Abstände zwischen den einzelnen Punkten führen dazu, dass der Spline den tatsächlichen Funktionsverlauf nur grob nachbilden kann [2]. Insbesondere in Bereichen mit stärkerer Krümmung sind deutliche Abweichungen erkennbar. Der Verlauf der Approximation wirkt teilweise zu flach und folgt der exakten Funktion nur eingeschränkt.

Mit 8 Stützstellen verbessert sich die Approximation deutlich. Durch die geringeren Abstände zwischen den Punkten kann der Spline den Verlauf der Funktion wesentlich besser erfassen [1]. Die Abweichungen zwischen exakter Funktion und Approximation sind sichtbar reduziert, treten jedoch weiterhin in Bereichen mit stärkerer Krümmung auf.

Bei 12 Stützstellen ist die Approximation nahezu identisch mit der exakten Funktion. Die Unterschiede sind visuell kaum noch erkennbar. Dies bestätigt empirisch, dass mit zunehmender Anzahl an Stützstellen die Approximationsqualität deutlich steigt [2].

Ein weiterer wichtiger Aspekt ist die Wahl der zugrunde liegenden Funktion. Durch die Kombination aus einer trigonometrischen Komponente und einem linearen Trend ergibt sich ein Funktionsverlauf mit unterschiedlichen Eigenschaften. Dadurch wird die Approximation anspruchsvoller und Unterschiede zwischen den Varianten werden deutlicher sichtbar. Dies zeigt, dass kubische Splines auch bei Funktionen mit unterschiedlichen Eigenschaften gute Approximationen liefern können.

Neben der Anzahl der Stützstellen spielt auch deren Verteilung eine zentrale Rolle [1], [2]. In dieser Arbeit wurden die Stützstellen gleichmässig verteilt, was eine einfache und nachvollziehbare Vorgehensweise darstellt. In Bereichen mit hoher Krümmung wäre jedoch eine dichtere Platzierung der Stützstellen sinnvoll, um die Approximation weiter zu verbessern. Eine adaptive Wahl der Stützstellen könnte somit zu einer effizienteren und genaueren Approximation führen [2].

Die Qualität der Approximation wurde sowohl visuell als auch durch eine einfache Fehleranalyse bewertet. Die grafische Darstellung ermöglicht eine intuitive Einschätzung, während die Berechnung des maximalen Fehlers eine quantitative Beurteilung erlaubt. Beide Ansätze ergänzen sich und tragen zu einem besseren Verständnis der Ergebnisse bei.

Zusammenfassend zeigt sich, dass kubische Splines eine leistungsfähige Methode zur Approximation nichtlinearer Funktionen darstellen und bereits mit einer moderaten Anzahl an Stützstellen eine hohe Genauigkeit erreicht werden kann [1], [2].
