# Kapitel 20 – Differentialrechnung II

# 1. Taylorpolynome und Taylorreihen

## Grundidee

Eine komplizierte Funktion wird in der Nähe eines Entwicklungspunktes x0 durch ein Polynom angenähert.

Je höher der Grad des Polynoms, desto genauer die Approximation.

---

## Taylorpolynom

Allgemeine Formel:

$$T_n(x)=\sum_{k=0}^{n}\frac{f^{(k)}(x_0)}{k!}(x-x_0)^k$$

Ausgeschrieben:

$$T_n(x)=f(x_0)+f'(x_0)(x-x_0)+\frac{f''(x_0)}{2!}(x-x_0)^2+\dots+\frac{f^{(n)}(x_0)}{n!}(x-x_0)^n$$

---

## Prüfungsablauf Taylorpolynom

### Schritt 1

Funktion bestimmen

Beispiel:

$$f(x)=sin(x)$$

---

### Schritt 2

Ableitungen bilden

$$f(x)=sin(x)$$

$$f'(x)=cos(x)$$

$$f''(x)=-sin(x)$$

$$f'''(x)=-cos(x)$$

---

### Schritt 3

Alle Ableitungen im Entwicklungspunkt einsetzen

$$Für x0=0:$$

$$f(0)=0$$

$$f'(0)=1$$

$$f''(0)=0$$

$$f'''(0)=-1$$

---

### Schritt 4

In Formel einsetzen

$$T_3(x)$$

$$=0+1·x+0·x²+(-1/6)x³$$

$$T_3(x)=x−x³/6$$

---

## Standard-Taylorreihen

### Exponentialfunktion

$$e^x$$

$$=1+x+x²/2!+x³/3!+x⁴/4!+...$$

---

### Sinus

$$\sin(x)$$

$$=x−x³/3!+x⁵/5!−x⁷/7!+...$$

---

### Kosinus

$$\cos(x)$$

$$=1−x²/2!+x⁴/4!−x⁶/6!+...$$

---

### Logarithmus

$$\ln(1+x)$$

$$=x−x²/2+x³/3−x⁴/4+...$$

$$für |x|<1$$

---

## Häufige Prüfungsfrage

Taylorreihe von sin(2x)

Einfach überall x durch 2x ersetzen:

$$\sin(2x)$$

$$=2x−(2x)³/3!+(2x)⁵/5!−...$$

---

## Restglied

Fehler:

$$R_n(x)=f(x)-T_n(x)$$

Merke:

Je näher x bei x0 liegt, desto kleiner wird der Fehler.

# 3. Lokale Extrema --> Ist das Lokale Extremum ein Minimum oder Maximum?

## Kandidaten finden

Immer: $f'(x)=0$ lösen

---

## Entscheidung mit zweiter Ableitung

### Minimum gefunden wenn $f''(x_0)>0$

$f'(x_0)=0$

$f''(x_0)>0$

---

### Maximum gefunden wenn $f''(x_0)<0$

$f'(x_0)=0$

$f''(x_0)<0$

---

## Beispiel

$f(x)=x²−4x+3$

$f'(x)=2x−4$

Nullstelle:

x=2

---

$f''(x)=2$

$f''(2)=2>0$

⇒ Minimum gefunden

---

## Prüfungsfalle

Wenn $f''(x0)=0$ --> also wenn die zweite Ableitung an der Stelle x_0 gleich 0 ergibt, darf man NICHT aufhören. Dann höhere Ableitungen prüfen.

---

## Prüfungsfalle 2

Die Prüfungsfrage geht davon aus, dass das Maximum oder Minimum bereits gegeben ist.

Wenn bei $x_0$ ein maximum bereits gegeben ist, dann gilt zwingend: $f'(x_0) = 0$ und $f''(x_0) \le 0$

Wenn bei $x_0$ ein minimum bereits gegeben ist, dann gilt zwingend: $f'(x_0) = 0$ und $f''(x_0) \ge 0$

Wenn bei der zweiten Ableitung 0 herauskommt, dann muss über die höhren Ableitungen entschieden werden, ob es sich um ein Maximum oder Minimum handelt. Siehe Punkt 5 dafür. Weil es könnte rauskommen, dass es ein Sattelpunkt ist und somit kein Maximum oder Minimu vorliegt.

---

# 4. Wendepunkte

## Kandidaten finden

$f''(x)=0$

lösen

---

## Entscheidung

Wenn

$f'''(x0)≠0$

⇒ Wendepunkt

---

## Sattelpunkt

Wenn zusätzlich

$f'(x0)=0$

dann:

Sattelpunkt

---

## Beispiel

$f(x)=x³$

$f'(x)=3x²$

$f''(x)=6x$

$f'''(x)=6$

---

$f''(0)=0$

$f'''(0)=6≠0$

⇒ Wendepunkt

---

$f'(0)=0$

⇒ Sattelpunkt

---

Bedenke: Ein Punkt kann sowohl Wendepunkt als auch Sattelpunkt gleichzeitig sein. Ein Sattelpunkt ist einfach ein spezieller Wendepunkt.

---

# 5. Höhere Ableitungen

Situation:

$f'(x0)=0$

$f''(x0)=0$

$f'''(x0)=0$

...

Erste nichtverschwindende Ableitung ist die n-te.

---

$n$ gerade

und positiv

⇒ Minimum

---

$n$ gerade

und negativ

⇒ Maximum

---

$n$ ungerade

⇒ Sattelpunkt und somit liegt kein Maximum oder Minimum vor.

---

## Klassiker

$f(x)=(x−3)^5$

erste vier Ableitungen verschwinden

Die 5. Ableitung $≠ 0$

$5$ ist ungerade

⇒ Sattelpunkt

# 6. Newton-Verfahren

## Idee und Bedeutung

Tangente verwenden. Nullstelle der Tangente wird neuer Näherungswert.

--> Achtung: Newton löst immer: $f(x)=0$

--> Du startest mit einem geschätzten Wert $x_0$. Dann setzt du diesen Wert in die Formel ein und bekommst $x_1$. Danach setzt du $x_1$ in die Formel ein und bekommst $x_2$, dann $x_3$, usw.

---

## Formel

$x_n=x_{n−1}−\frac{f(x_{n−1})}{f'(x_{n−1})}$

oder einfacher ausgedrückt: $x_{neu}=x_{alt}−\frac{f(x_{alt})}{f'(x_{alt})}$

---

## Prüfungsbeispiel

$\sqrt{2}$ berechnen

$f(x)=x^2−2$

$f'(x)=2x$

Start:

$x_0=1$

---

In Formel einfügen:

$x_{1} = 1 − \frac{1^2−2}{2 * 1}$

$=1.5$

---

$x_{2} = 1.5 − \frac{1.5^2−2}{2 * 1.5}$

$≈1.41667$

---

$x_{3}$

$≈1.41422$

---

$x_{4}$

$≈1.41421$

---

Ergebnis:

$\sqrt{2}≈1.41421$

---

## Hinweis

Wie weit man gehen soll, hängt von der Aufgabenstellung ab. Du hörst auf, wenn eine dieser Bedingungen erfüllt ist:

- Die Aufgabe sagt es ausdrücklich: "Führen Sie 3 Iterationen durch" --> also rechnen bis $x_3$
- Die gewünschte Genauigkeit ist erreicht: Zum Beispiel "auf 3 Nachkommastellen genau". --> dann rechnest du weiter, bis sich die ersten 3 Nachkommatellen nicht mehr ändern.
- Zwei aufeinanderfolgende Werte sind praktisch gleich: Wenn $x_3$ und $x_4$ fast gleich sind, kannst du aufhören.

---

## Prüfungsfalle

Die Prüfung kann eine Gleichung als Ausgangspunkt vorgeben. In diesem Fall musst du zuerst die Gleichung so umstellen, dass sie die Form $f(x)=0$ hat, bevor du das Newton-Verfahren anwenden kannst. Beispiel: vorgegeben ist die Gleichung $x^2=2$. Du musst sie umstellen zu $f(x)=x^2−2=0$, damit du das Newton-Verfahren anwenden kannst. So erhälst du $f(x)$, was in diesem Fall $f(x)=x^2−2$ ist, und dann kannst du die Ableitung $f'(x)=2x$ berechnen und das Newton-Verfahren anwenden.

Wenn kein Startwert (x_0) vorgegeben wird, dann kannst du einen beliebigen Startwert wählen. Es ist oft sinnvoll, einen Startwert zu wählen, der nahe an der erwarteten Lösung liegt, um die Konvergenz zu beschleunigen. Zum Beispiel, wenn du die Quadratwurzel von 2 berechnen möchtest, könntest du mit einem Startwert von 1 beginnen, da 1 nahe an $\sqrt{2}$ liegt.

--> Den Startwert, der nahe an der Lösung liegt, findest du, indem du die Gleichung löst. Also $x^2=2$: hier würdest du die Wurzel ziehen, also ergibt das $x=\sqrt{2}$. Da $\sqrt{2}$ ungefähr 1.41421 ist, könntest du einen Startwert von 1.4 oder 1.5 wählen, um die Konvergenz zu beschleunigen.

--> Falls die Gleichung zu kompliziert ist, einfach mit Zahlen experimentieren, bis du einen Startwert findest, der nahe an der Lösung liegt. Zum Beispiel: $x^5 + x^3 + x -2 = 0$: Mit 0 probieren ergibt $-2$, mit 1 probieren ergibt 1, also liegt die Lösung irgendwo zwischen 0 und 1. Dann könntest du mit einem Startwert von 0.5 beginnen.

# 7. Sekantenverfahren

## Unterschied zu Newton

Newton:

- benötigt Ableitung
- schneller

Sekante:

- benötigt keine Ableitung
- zwei Startwerte

---

## Formel

$$x_n = x_{n−1} - f(x_{n−1}) \cdot \frac{x_{n−1}-x_{n−2}}{f(x_{n−1}) - f(x_{n−2})}$$

oder einfacher ausgedrückt:

$$x_{neu} = x_{1} - f(x_{1}) \cdot \frac{x_{1}-x_{0}}{f(x_{1}) - f(x_{0})}$$

---

## Beispiel

Wir suchen die positive Nullstelle von $f(x)=x^2−4$. Wir tun so, als hätten wir keine Ableitung zur Verfügung, also verwenden wir das Sekantenverfahren.

Startwerte:

- $x_0=1$
- $x_1=3$

Funtionswerte:

- $f(1) = 1^2 - 4 = -3$
- $f(3) = 3^2 - 4 = 5$

Einsetzen in Formel:
$$x_2 = x_1 - f(x_1) * \frac{x_1 - x_0}{f(x_1) - f(x_0)} = 3 - 5 \cdot \frac{3-1}{5 - (-3)} = 1.75$$

Also $x_2 = 1.75$

Jetzt haben wir $x_1$ und $x_2$ als neue Startwerte:

- $x_1=3$
- $x_2=1.75$

Funtionswerte:

- $f(3) = 5$
- $f(1.75) = 1.75^2 - 4 = -0.9375$

Einsetzen in Formel:
$$x_3 = x_2 - f(x_2) * \frac{x_2 - x_1}{f(x_2) - f(x_1)} = 1.75 - (-0.9375) \cdot \frac{1.75 - 3}{-0.9375 - 5} = 1.94737$$

Also $x_3 ≈ 1.94737$

Und somit schon näher bei 2.
