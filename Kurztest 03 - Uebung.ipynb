# Typische Prüfungsaufgaben – Integralrechnung

---

# Aufgabe 1 — Unbestimmtes Integral mit Linearität

## Aufgabe

Berechnen Sie:

$$
\int (3x^2 - 4x + 5)\,dx
$$

---

## Schritt 1 — Aufgabentyp erkennen

Keine Grenzen vorhanden.

→ Unbestimmtes Integral
→ Stammfunktion gesucht

Außerdem besteht der Integrand aus mehreren Summanden.

→ Linearität verwenden

---

## Schritt 2 — Integral aufteilen

$$
\int (3x^2 - 4x + 5)\,dx
=
\int 3x^2\,dx
-
\int 4x\,dx
+
\int 5\,dx
$$

Konstanten nach vorne ziehen:

$$
=
3\int x^2\,dx
-
4\int x\,dx
+
5\int 1\,dx
$$

---

## Schritt 3 — Einzelteile integrieren

Potenzregel:

$$
\int x^2\,dx = \frac{x^3}{3}
$$

$$
\int x\,dx = \frac{x^2}{2}
$$

Konstante:

$$
\int 1\,dx = x
$$

---

## Schritt 4 — Einsetzen

$$
=
3\cdot\frac{x^3}{3}
-
4\cdot\frac{x^2}{2}
+
5x
+
C
$$

---

## Schritt 5 — Vereinfachen

$$
=
x^3 - 2x^2 + 5x + C
$$

---

## Typische Prüfungsfehler

- Integrationskonstante \(C\) vergessen
- Exponenten nicht um 1 erhöhen
- Konstanten nicht korrekt behandeln

# Aufgabe 2 — Bestimmtes Integral

## Aufgabe

Berechnen Sie:

$$
\int_1^3 (2x+1)\,dx
$$

---

## Schritt 1 — Aufgabentyp erkennen

Grenzen vorhanden.

→ Bestimmtes Integral

Es gilt der Hauptsatz:

$$
\int_a^b f(x)\,dx = F(b)-F(a)
$$

---

## Schritt 2 — Stammfunktion bilden

Gegeben:

$$
f(x)=2x+1
$$

Stammfunktion:

$$
F(x)=x^2+x
$$

Denn:

$$
\int 2x\,dx = x^2
$$

und:

$$
\int 1\,dx = x
$$

---

## Schritt 3 — Grenzen einsetzen

Obere Grenze:

$$
F(3)=3^2+3=12
$$

Untere Grenze:

$$
F(1)=1^2+1=2
$$

---

## Schritt 4 — Subtrahieren

$$
12-2=10
$$

---

## Endresultat

$$
\int_1^3 (2x+1)\,dx = 10
$$

---

## Typische Prüfungsfehler

- Grenzen direkt in \(f(x)\) statt in \(F(x)\) einsetzen
- Untere Grenze nicht abziehen

# Aufgabe 3 — Substitution

## Aufgabe

Berechnen Sie:

$$
\int (3x+2)^4\,dx
$$

---

## Schritt 1 — Aufgabentyp erkennen

Es gibt:
- eine äußere Funktion: Potenz
- eine innere Funktion: \(3x+2\)

→ Typischer Fall für Substitution

---

## Schritt 2 — Substitution wählen

Setze:

$$
u=3x+2
$$

Ableiten:

$$
du = 3\,dx
$$

Nach \(dx\) auflösen:

$$
dx=\frac13 du
$$

---

## Schritt 3 — Integral umschreiben

Aus:

$$
(3x+2)^4
$$

wird:

$$
u^4
$$

und:

$$
dx=\frac13 du
$$

Damit:

$$
\int (3x+2)^4\,dx
=
\int u^4\cdot\frac13\,du
$$

Konstante nach vorne ziehen:

$$
=
\frac13\int u^4\,du
$$

---

## Schritt 4 — Integrieren

Potenzregel:

$$
=
\frac13\cdot\frac{u^5}{5}+C
$$

$$
=
\frac{u^5}{15}+C
$$

---

## Schritt 5 — Rücksubstitution

\(u\) wieder ersetzen:

$$
=
\frac{(3x+2)^5}{15}+C
$$

---

## Typische Prüfungsfehler

- \(dx\) nicht korrekt ersetzen
- Rücksubstitution vergessen
- Nur \(u\) substituieren, aber nicht \(du\)

---

# Allgemeines Vorgehensmodell für Prüfungen

| Muster | Methode |
|---|---|
| Summe mehrerer Terme | Linearität |
| Grenzen vorhanden | Hauptsatz |
| Innere Funktion vorhanden | Substitution |
| Produkt verschiedener Funktionen | Partielle Integration |
